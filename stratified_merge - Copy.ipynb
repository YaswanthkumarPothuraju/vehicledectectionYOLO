# Count images, labels, and objects in each split

```python
from pathlib import Path
from collections import Counter

DATASET = Path(
    "/content/drive/MyDrive/VEHDATA/datasetweused/merged_vehicle_dataset_stratified_70_20_10"
)

CLASS_NAMES = [
    "bicycle",
    "bus",
    "car",
    "bike",
    "auto",
    "truck",
    "van",
    "person",
    "concrete_mixer",
    "tractor",
    "vehicle_truck"
]

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp", ".tif", ".tiff"
}

if not DATASET.exists():
    raise FileNotFoundError(f"Dataset not found: {DATASET}")

total_images = 0
total_objects = Counter()

for split in ["train", "valid", "test"]:
    image_dir = DATASET / split / "images"
    label_dir = DATASET / split / "labels"

    if not image_dir.exists() or not label_dir.exists():
        print(f"\nWARNING: Missing images or labels directory in {split}")
        continue

    images = [
        p for p in image_dir.rglob("*")
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    labels = list(label_dir.rglob("*.txt"))
    object_counts = Counter()
    images_per_class = Counter()
    empty_labels = 0

    for label_file in labels:
        lines = label_file.read_text(
            encoding="utf-8"
        ).splitlines()

        valid_class_ids_in_image = set()

        if not any(line.strip() for line in lines):
            empty_labels += 1

        for line in lines:
            parts = line.strip().split()

            if not parts:
                continue

            try:
                class_id = int(float(parts[0]))
            except (ValueError, IndexError):
                continue

            if 0 <= class_id < len(CLASS_NAMES):
                object_counts[class_id] += 1
                valid_class_ids_in_image.add(class_id)

        for class_id in valid_class_ids_in_image:
            images_per_class[class_id] += 1

    total_images += len(images)
    total_objects.update(object_counts)

    print(f"\n{'=' * 65}")
    print(f"{split.upper()} SPLIT")
    print(f"{'=' * 65}")
    print(f"Images:                 {len(images)}")
    print(f"Label files:            {len(labels)}")
    print(f"Empty label files:      {empty_labels}")
    print(f"Total annotated objects: {sum(object_counts.values())}")

    print(f"\n{'ID':<5}{'Class':<20}{'Objects':>12}{'Images':>12}")
    print("-" * 49)

    for class_id, class_name in enumerate(CLASS_NAMES):
        print(
            f"{class_id:<5}"
            f"{class_name:<20}"
            f"{object_counts[class_id]:>12}"
            f"{images_per_class[class_id]:>12}"
        )

print(f"\n{'=' * 65}")
print("DATASET SUMMARY")
print(f"{'=' * 65}")
print("Total images:", total_images)
print("Total objects:", sum(total_objects.values()))
print("Number of classes:", len(CLASS_NAMES))
```
